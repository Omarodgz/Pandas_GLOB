# Librerías

Cargando las bibliotecas que necesitamos 


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
```

## Mostrando todos los archivos de "/datos"

La salida serán los archivos en forma de lista

```python
# establecer ruta de búsqueda  
path = 'datos'
files = glob.glob(path + '/*') # mostrar en forma de lista todos los archivos que se encuentran en esta carpeta
files
```

## Mostrando  los archivos  **.txt** de "/datos"

La salida serán los archivos .txt en forma de lista

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos txt  
path = 'datos'
files = glob.glob(path + '/*.txt') # buscar todos los archivos de tipo .txt
files
```

## Mostrando  los archivos .csv de "/datos"

La salida serán los archivos .csv en forma de lista

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/*.csv')
files
```

## Concatenando todos los archivos .csv para formar un dataframe


```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/*.csv')
print(files)
# crear una lista vacía para almacenar  los dataframes
lista_dfs = []
# recorrer la lista de archivos para crear la lista de archivos 
for file in files:
    temp_df = pd.read_csv(file)
    print(temp_df.shape)
    lista_dfs.append(temp_df)

df = pd.concat(lista_dfs, axis = 0)
df.head()
print('Final: ', df.shape)
# concatenación
```

## Concatenando los archivos .csv del primer semestre 

El comodín * representa desde nada, hasta cualquier cantidad de caracteres y dígitos

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/semestre1*.csv') # También puede ser semestre1?.csv
print(files)
# crear una lista vacía para almacenar  los dataframes
lista_dfs = []
# recorrer la lista de archivos para crear la lista de archivos 
for file in files:
    temp_df = pd.read_csv(file)
    print(temp_df.shape)
    lista_dfs.append(temp_df)

df = pd.concat(lista_dfs, axis = 0)
df.head()
print('Final: ', df.shape)
# concatenación
```

## Concatenando los archivos .csv del primer y segundo semestre 

El comodín **[ ]**  representa un rango,  ya sea de caracteres o de números. 

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/semestre[1-2]*') # se usa [] para un rango de números, también puede ser semestre[1-2]?.csv
print(files)
# crear una lista vacía para almacenar  los dataframes
lista_dfs = []
# recorrer la lista de archivos para crear la lista de archivos 
for file in files:
    temp_df = pd.read_csv(file)
    print(temp_df.shape)
    lista_dfs.append(temp_df)

df = pd.concat(lista_dfs, axis = 0)
df.head()
print('Final: ', df.shape)
# concatenación
```

## Concatenando los archivos .csv de los grupos "a"
El comodín **?** Representa un único carácter.

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/semestre?a*') 
print(files)
# crear una lista vacía para almacenar  los dataframes
lista_dfs = []
# recorrer la lista de archivos para crear la lista de archivos 
for file in files:
    temp_df = pd.read_csv(file)
    print(temp_df.shape)
    lista_dfs.append(temp_df)

df = pd.concat(lista_dfs, axis = 0)
df.head()
print('Final: ', df.shape)
# concatenación
```

## Concatenando los archivos .csv de los grupos "a" y "b"

El comodín **[ ]**  representa un rango,  ya sea de caracteres o de números. 

```python
# establecer ruta de búsqueda  
# aquí queremos buscar archivos csv  
path = 'datos'
files = glob.glob(path + '/semestre?[a,b]*') # '/semestre?[a,b].csv'
print(files)
# crear una lista vacía para almacenar  los dataframes
lista_dfs = []
# recorrer la lista de archivos para crear la lista de archivos 
for file in files:
    temp_df = pd.read_csv(file)
    print(temp_df.shape)
    lista_dfs.append(temp_df)

df = pd.concat(lista_dfs, axis = 0)
df.head()
print('Final: ', df.shape)
# concatenación
```
